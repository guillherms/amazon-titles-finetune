```python
from unsloth import FastLanguageModel
import torch

repo_id = "guillherms/llama-3.2-3b-amazon-titles-100k-lora"

model, tokenizer = FastLanguageModel.from_pretrained(
    repo_id,
    max_seq_length=2048,
    dtype=None,
    load_in_4bit=True,
)

FastLanguageModel.for_inference(model)

```

```python
import random, time, pandas as pd

instructions = [
    "Describe this product",
    "Explain what this product is",
    "Provide details about this product",
    "Give a description of the product type",
    "Summarize the type of this product",
    "Describe a product by title"
]

products = [
    "Men's Cotton T-Shirt with Minimalist Logo",
    "Women's Floral Summer Dress with Side Slit",
    "Wireless Noise Cancelling Headphones",
    "Hydrating Face Serum with Hyaluronic Acid",
    "Air Fryer 5L Digital Touch Display",
    "Smartwatch with Heart Rate and Sleep Tracking",
    "Round Plush Calming Bed for Dogs",
    "Gaming Chair Reclining with Footrest",
    "USB-C to Lightning Nylon Cable 1.5m",
    "Organic Argan Oil Hair Treatment 200ml"
]


results = []

for product in products:
    instruction = random.choice(instructions)
    prompt = f"Instruction: {instruction}\nInput: {product}\nOutput:"

    inputs = tokenizer(prompt, return_tensors="pt").to(model.device)

    start = time.time()
    outputs = model.generate(
        **inputs,                 # tokenized prompt
        max_new_tokens = 120,     # maximum number of tokens to generate (response length)
        temperature = 0.7,        # controls creativity (lower = precise, higher = more diverse)
        top_p = 0.9,              # nucleus sampling: keep only most likely tokens
        do_sample = True,         # enables random sampling (required for variation)
        repetition_penalty = 1.2, # penalizes repeated words or phrases
        no_repeat_ngram_size = 3, # prevents exact repetition of 3-word sequences
    )
    elapsed = time.time() - start
    text = tokenizer.decode(outputs[0], skip_special_tokens=True).strip()

    results.append({
        "instruction": instruction,
        "product": product,
        "description": text,
        "elapsed_sec": round(elapsed, 2)
    })

    print(f"\n🧭 {instruction}")
    print(f"🛒 {product}")
    print(f"📝 {text[:400]}{'...' if len(text)>400 else ''}")
    print(f"⚡ Time: {elapsed:.2f}s")
```

```python
df = pd.DataFrame(results)
df["word_count"] = df["description"].apply(lambda x: len(x.split()))
df["repetition_ratio"] = df["description"].apply(lambda x: len(x.split()) / len(set(x.split())))

display(df.head())
df.to_csv("alpaca_product_model_results.csv", index=False)
print("\n✅ Saved results to alpaca_product_model_results.csv")

# Optional summary
print("\n📊 Average tokens per description:", round(df.word_count.mean(), 1))
print("📊 Avg repetition ratio:", round(df.repetition_ratio.mean(), 2))
```
